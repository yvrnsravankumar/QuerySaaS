Oracle Fusion BI Publisher to DuckDB using the QuerySaaS library.


Before installing, place the complete XDRZ Base64 archive in `src/querysaas/xdrz_payload.py`, then build or install the package.

```python
from getpass import getpass
from querysaas import connect
```

```python
FUSION_URL = "https://your-host.oraclecloud.com"
FUSION_USERNAME = "your.username"
FUSION_PASSWORD = getpass("Oracle Fusion password: ")

con = connect(
    "oracle_fusion",
    url=FUSION_URL,
    username=FUSION_USERNAME,
    password=FUSION_PASSWORD,
)
```

```python
df = con.executequery("SELECT * FROM dual", all_varchar=True)
display(df)
```

```python
result = con.copy2dd_parallel(
    table_name="hz_geographies",
    primary_key="GEOGRAPHY_ID",
    count=5000,
    max_workers=4,
    duckdb_path="querysaas.duckdb",
    all_varchar=True,
)
print(result)
```

```python
con.close()
```
